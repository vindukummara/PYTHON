 # Match case statement

```python
# Match case statement is used to replace multiple if-else conditions.
# Match case is similar to Switch case in C & C++.
# Match case is more advance than the Switch case.
# Match case is having advance features like handilng of data structure and conditions.
```

```python
# EXample-1 : structure of match case

num = int(input("Enter a number"))
match num:
    case 10:
        print("Hello")
    case 20:                                  # it is variable value like 10,20,30
        print("World")
    case _:
        print("something else")
```

```python
# Example-2 : match case with data structure
data = [10,20,30]

match data:
    case [1,2,3]:              # DATA structure [10,20,30]
        print("Hello-1")
    case [30,20,10]:
        print("Hello-2")
    case [10,20,30]:
        print("Hello-3")
    case _:
        print("Nothing Match")
```

```python
# Example -3: Match case using condition

num = int(input("Enter a number"))

match num: 
    case x if x > 0:
        print("It is positive number")           # it is conditon if x>0:
    case y if y < 0:
        print("It is negative number")
    case _:
        print("It is ZERO")
    
```

## 1. Function overloading

```python
# code to overload the function in C++ or Java
# C++, Java type of only polymoriphism is NOT applicable in python directly

from multipledispatch import dispatch


@dispatch(int)
def fun(var1):
    print(var1)
@dispatch(int,int)    
def fun(var1,var2):
    print(var1, var2)
@dispatch(int,int,int)    
def fun(var1,var2,var3):
    print(var1, var2,var3)
fun(10)          
fun(10, 20)
fun(10,20,30)
```

```python
# Overloading function using "Defalut parameter"
def fun(var1, var2 = None):   # here var2 = None is the defalut parameter
    if var2 is not None:
        print(var1, var2)
    else:
        print(var1)
fun(10)
fun(10, 20)
```

#### Arthematic operater overloading


```python
# Target : Overload the arthimatic operator for user defined objects.

class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
d1 = Dosa("Masala", 80)
d2 = Dosa("Ghee Karam", 150)

total_price = d1 + d2
print(total_price)
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price

    # __add__() function is helping to overload the + operator
    def __add__(self, obj2):
        return self.Price + obj2.Price
        
d1 = Dosa("Masala", 80)
d2 = Dosa("Ghee Karam", 150)

total_price = d1 + d2
print(total_price)
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
     # __add__() function is helping to overload the + operator
    def __add__(self, obj2):
        return self.Price + obj2.Price 
     # __sub__() function is helping to overload the - operator
    def __sub__(self, obj2):
        return self.Price - obj2.Price 

     # __mul__() function is helping to overload the - operator
    def __mul__(self, obj2):
        return self.Price * obj2.Price 

     # __truediv__() function is helping to overload the - operator
    def __truediv__(self, obj2):
        return self.Price / obj2.Price 
     # __floordiv__() function is helping to overload the - operator
    def __floordiv__(self, obj2):
        return self.Price  // obj2.Price 
    # __mod__() function is helping to overload the - operator
    def __mod__(self, obj2):
        return self.Price  % obj2.Price 
    # __pow__() function is helping to overload the - operator
    def __pow__(self, obj2):
        return self.Price ** obj2.Price 
d1 = Dosa("Masala", 80)
d2 = Dosa("Ghee Karam", 150)

total_price = d1 + d2  #__add__()
print(total_price)
print(d2 - d1)        #__sub__()
print(d2 * d1)        #__mul__()
print(d2 / d1)        #__truediv__()
print(d2 // d1)       #__floordiv__()
print(d2 % d1)        #__mod__()
print(d2 ** d1)       #__pow__()
```
